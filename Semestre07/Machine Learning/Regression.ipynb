Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
```

1- DataSet :

```python
a = 0.5 ; b = 0.4
x = np.linspace(0, 10, 50)
y = 2 * x + 1 + np.random.randn(50) * 1.5 
fig , ax = plt.subplots()
ax.grid(6,alpha=0.5)
ax.scatter(x,y,c="crimson",edgecolors='k')
```

2 - Modèle linéaire :

```python
ax.plot(x, a * x + b , label = "Initial Line")
ax.legend()
ax.set_xlabel('x', loc= 'right')
ax.set_ylabel('y', loc = 'top')
ax.set_title("Régression")
fig
```

3- Fonction Cout :

```python
#   J(a,b)
def fonction_cout(x,y,a,b,m):
    s = 0
    for i in range(m):
        s += (a * x[i] + b - y[i])**2
    return s / (2*m)

# Version Plus Simple:
def fonction_cout_simp(x,y,a,b,m):    
    return np.sum( ( a * x + b - y ) **2  / (2 * m))

print(f"J({a},{b})= {fonction_cout(x,y,0.5,0.4,50):.3f}.")
```
