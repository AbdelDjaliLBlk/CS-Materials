```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

1. DataSet:

```python
x,y = make_regression(n_samples=100,n_features=1,noise=10)
plt.scatter(x,y)
```

```python
# Shape of x,y :
y = y.reshape(y.shape[0],1)
print(x.shape)
print(y.shape)
```

```python
# Matrice X
X = np.hstack((x, np.ones(x.shape)))
X.shape
```

```python
# Theta 
theta = np.random.randn(2,1)
theta
```

2. Modèle:

```python
def model(X,theta):
    return X.dot(theta) 
```

```python
plt.scatter(x,y)
plt.plot(x,model(X,theta) , color='r')
```

3. Fonction cout:

```python
def cost_function(X, y, theta):
    m = len(y)
    return 1/(2*m) * np.sum((model(X,theta) - y)**2)
```

```python
float(cost_function(X,y,theta))
```

4. Gradient & Descente de Gradient :

```python
def grad(X,y, theta):
    m = len(y)
    return 1/m * X.T.dot(model(X,theta) - y)
```

```python
def gradien_descent(X,y,theta,alpha,n_iteration):
    cost_history = np.zeros(n_iteration)
    for _ in range(n_iteration):
        theta = theta - alpha * grad(X,y,theta)
        cost_history[_] = cost_function(X,y,theta)
    return theta,cost_history
```

5. Entrainement:

```python
N = 1000
theta_final ,cost_history = gradien_descent(X,y,theta,0.01,N)
```

```python
prediction = model(X,theta_final)
plt.scatter(x,y)
plt.plot(x,prediction, c = 'r')
```

6. Courbe d'apprentissage:

```python
plt.plot(range(N), cost_history)
```

7. Coefficient de Determination

```python
def coeff_determination(y,pred):
    U = ((y - pred)**2).sum()
    V = ((y - y.mean())**2).sum()
    return 1 - U/V
```

```python
float(coeff_determination(y,prediction))
```
