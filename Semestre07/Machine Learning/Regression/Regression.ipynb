                                Linear Regression

```python
import numpy as np
import matplotlib.pyplot as plt
```

1- DataSet :

```python
#------------------
a = 0.5 ; b = 0.4
x = np.linspace(0, 10, 50)
y = 2 * x + 1 + np.random.randn(50) * 1.2
fig , ax = plt.subplots(figsize=(10,6))
ax.grid(6,alpha=0.5)
ax.scatter(x,y,c="crimson",edgecolors='k')
```

2 - Modèle linéaire :

```python
ax.plot(x, a * x + b , label = f"Initial Line (a = {a}, b = {b})")
ax.legend()
ax.set_xlabel('x', loc= 'right')
ax.set_ylabel('y', loc = 'top')
ax.set_title("Régression Linéaire")
fig
```

3- Fonction Cout :

```python
#   J(a,b)
def fonction_cout(x,y,a,b):
    m = len(x)
    s = 0
    for i in range(m):
        s += (a * x[i] + b - y[i])**2
    return s / (2*m)

# Version Plus Simple:
def fonction_cout_simp(x,y,a,b):
    m = len(x)    
    return np.sum( ( a * x + b - y ) **2  / (2 * m))

print(f"Old J(a,b)= {fonction_cout(x,y,a,b):.3f}.")
```

4- Descente de Gradient :

```python
# Algorithme de Minimisation-
def gradient_descent(x,y,a,b,alpha,iterations):
    m = len(x)
    for _ in range(iterations):
        dy = a * x + b - y
        grad_a = np.sum(x * dy) / m
        grad_b = np.sum(dy) / m
        a = a - alpha * grad_a
        b = b - alpha * grad_b   
    return a,b   
#------------------
alpha = 0.01
N = 1000
a,b = gradient_descent(x,y,a,b,alpha,N)
#------------------
ax.plot(x, a * x + b , label= f"Solution Line (a = {a:.2f}, b = {b:.2f},α = {alpha} , n = {N})" , c = 'r')
ax.legend() 
#------------------
print(f"A = {b:.3f}.\nB = {a:.3f}")
print(f"New J(a,b)= {fonction_cout(x,y,a,b):.3f}.")
fig
```
