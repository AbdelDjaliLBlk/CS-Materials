```python
import numpy as np
import matplotlib.pyplot as plt
```

1- DataSet :

```python
a = 0.5 ; b = 3 ; c = 1/2
x = np.linspace(0,10,50)
y =  np.random.rand(50)* x**2 + np.random.rand(50) * x + np.random.rand(50) 
fig ,ax = plt.subplots(figsize=(10,6))
ax.scatter(x,y,c='crimson',edgecolors="k", label="Data")
ax.grid(alpha=0.5)
```

2- Modèle Polynomial:

```python
ax.plot(x,a * x**2 + b*x +c,label=f"{a}$x^2$ + {b}x +{c}")
ax.legend()
ax.set_xlabel('x', loc= 'right')
ax.set_ylabel('y', loc = 'top')
ax.set_title("Régression Polynomiale")
fig
```
